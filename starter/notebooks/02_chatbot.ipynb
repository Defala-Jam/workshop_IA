# 02. Um chatbot com histórico

Para continuar uma conversa, precisamos guardar as mensagens e enviá-las novamente ao modelo. A função `chat` abaixo adiciona a pergunta à lista, chama o modelo e guarda a resposta. Execute as duas perguntas e veja se o tutor recupera o tema informado.

```python
from langchain_ollama import ChatOllama
from langchain.messages import SystemMessage, HumanMessage, AIMessage

model = ChatOllama(base_url="http://127.0.0.1:11434", model="qwen3.5:4b", temperature=0, reasoning=False)
messages = [SystemMessage(content="Você é um tutor de ciências de ensino fundamental.")]
```

```python
def chat(user_text: str) -> str:
    messages.append(HumanMessage(content=user_text))
    answer = model.invoke(messages)
    messages.append(answer)
    return str(answer.content)

print(chat("Hoje vou estudar ecossistemas."))

print("\n","=="*20,"\n")

print(chat("Qual tema eu disse que estou estudando?"))
```

## Veja o que foi guardado

Execute a célula e observe a ordem das mensagens: instrução, pergunta, resposta e assim por diante. Esse histórico está na lista `messages` e será perdido se você reiniciar o kernel.

```python
for i, message in enumerate(messages, start=1):
    print(f"Mensagem {i}:")
    print(message.content)
    print("\n" + "==" * 20 + "\n")
```

## Experimento: retire o histórico

A chamada abaixo mantém a instrução do tutor e a última pergunta, mas não inclui a mensagem que informa o tema. Compare a resposta com a anterior. Sem esse contexto, o modelo não tem como saber qual tema você escolheu.

```python
fresh_context = [
    SystemMessage(content="Você é um tutor de ciências. Responda em português de forma breve."),
    HumanMessage(content="Qual tema eu disse que estou estudando?"),
]
print(model.invoke(fresh_context).content)
```
